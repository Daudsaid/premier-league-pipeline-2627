```python
from pathlib import Path

import pandas as pd

cwd = Path.cwd()
project_root = cwd if (cwd / "data" / "landed").exists() else cwd.parent

landed_files = sorted((project_root / "data" / "landed").glob("*.csv"))
print(landed_files)
latest = landed_files[-1]
print(latest)

df = pd.read_csv(latest, encoding="utf-8-sig")
df.shape
```

```python
match_cols = [
    "Div", "Date", "Time", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR",
    "HTHG", "HTAG", "HTR", "Referee", "HxG", "AxG",
    "HS", "AS", "HST", "AST", "HF", "AF", "HC", "AC", "HY", "AY", "HR", "AR",
]
odds_cols = [c for c in df.columns if c not in match_cols]
len(odds_cols), odds_cols
```

```python
from collections import Counter

suffixes = ["CAHA", "CAHH", "AHA", "AHH", "C>2.5", "C<2.5", ">2.5", "<2.5", "CH", "CD", "CA", "H", "D", "A"]

def strip_suffix(col):
    if col in ("AHh", "AHCh"):
        return col
    for suf in sorted(suffixes, key=len, reverse=True):
        if col.endswith(suf):
            return col[: -len(suf)]
    return col

prefixes = Counter(strip_suffix(c) for c in odds_cols if c not in ("AHh", "AHCh"))
prefixes
```
